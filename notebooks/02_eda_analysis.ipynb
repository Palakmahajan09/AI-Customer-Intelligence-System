```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_style("whitegrid")

df = pd.read_csv('../data/processed/cleaned_data.csv')

df.head()
df.shape
df.columns
df.describe()

# graphs - top 10 customers
import matplotlib.pyplot as plt
top_customers = df.groupby('customer_id')['payment_value'].sum().sort_values(ascending=False).head(10)

top_customers.plot(kind='bar')

plt.title("Top 10 Customers By Revenue")
plt.xlabel("Customer ID")
plt.ylabel("Revenue")
plt.xticks(rotation=45)

plt.show()

# payment type
df['payment_type'].value_counts().plot(kind='bar')

plt.title("Payment Type Distribution")
plt.xlabel("Payment Type")
plt.ylabel("Count")

plt.show()

# monthly sales
df['order_purchase_timestamp'] = pd.to_datetime(df['order_purchase_timestamp'])

df['month'] = df['order_purchase_timestamp'].dt.to_period('M')

monthly_sales = df.groupby('month')['payment_value'].sum()

monthly_sales.plot()
plt.title("Monthly Revenue Trend")
plt.xlabel("Month")
plt.ylabel("Revenue")
plt.show()

# Top product category
top_categories = df['product_category_name'].value_counts().head(10)

top_categories.plot(kind='bar')
plt.title("Top Product Categories")
plt.xlabel("Category")
plt.ylabel("Count")
plt.xticks(rotation=45)
plt.show()



```

## Statistical Summary Insights

- The average payment value is around 172.85, while the maximum goes up to 13664.08, indicating presence of high-value transactions (possible outliers).

- The average product price is 120.82, but the wide range between minimum and maximum suggests significant variation in product pricing.

- Freight value has a mean of 20.04, showing additional cost impact on total purchase value.

- Most customers make payments in 1–3 installments, as seen from the median (2 installments).

- Product weight and dimensions vary significantly, indicating a diverse product catalog.

- Some fields contain missing values (difference in count), which were handled during data cleaning.

- The distribution shows skewness in payment value and price, indicating that few high-value transactions dominate overall revenue.
