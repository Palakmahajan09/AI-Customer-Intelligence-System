```python
import pandas as pd

customers = pd.read_csv('../data/raw/olist_customers_dataset.csv')
orders = pd.read_csv('../data/raw/olist_orders_dataset.csv')
products = pd.read_csv('../data/raw/olist_products_dataset.csv')
payments = pd.read_csv('../data/raw/olist_order_payments_dataset.csv')

customers.info()
customers.isnull().sum()
customers.nunique()

orders.info()
orders.isnull().sum()
orders.nunique()

products.info()
products.isnull().sum()
products.nunique()

payments.info()
payments.isnull().sum()
payments.nunique()

customers.shape
orders.shape
products.shape
payments.shape
products['product_category_name'] = products['product_category_name'].fillna('Unknown')
orders['order_delivered_customer_date'] = orders['order_delivered_customer_date'].ffill()
order_items = pd.read_csv('../data/raw/olist_order_items_dataset.csv')
order_items.head()
df = orders.merge(customers, on='customer_id', how='left')
df = df.merge(payments, on='order_id', how='left')
df = df.merge(order_items, on='order_id', how='left')
df = df.merge(products, on='product_id', how='left')
df.head()
df.shape

df.to_csv('../data/processed/cleaned_data.csv', index=False)
```

## Data Cleaning & Merging

- Handled missing values in product category and order delivery dates
- Combined multiple tables using keys like customer_id and order_id
- Created a unified dataset for further analysis

## Table Relationships

- customers → orders using customer_id
- orders → payments using order_id
- products → linked through order_items (to be used later)
